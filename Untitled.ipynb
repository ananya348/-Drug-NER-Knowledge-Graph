Cell 4: Train–Test Split on Excel Rows:

```python
df_train, df_test = train_test_split(df, test_size=0.2, random_state=42)
print("✅ Train rows:", len(df_train))
print("✅ Test rows :", len(df_test))
```

Cell 5: Helper Functions and Sentence Templates:

```python
def split_semicolon(x):
    if pd.isna(x):
        return []
    return [t.strip() for t in str(x).split(";") if t.strip()]

TEMPLATES = [
    "{DRUG} is used to treat {CANCERS} cancer and may cause {SIDEFX}.",
    "{DRUG} is indicated in {CANCERS} and can lead to {SIDEFX}.",
    "Treatment with {DRUG} for {CANCERS} has been associated with {SIDEFX}.",
    "{DRUG} therapy in {CANCERS} patients may result in {SIDEFX}.",
    "In {CANCERS}, {DRUG} may cause adverse effects such as {SIDEFX}.",
    "{SIDEFX} has been reported in patients receiving {DRUG} for {CANCERS}."
]
```

 Build NER Training and Test Samples:

```python
def build_training_data(df_in, templates, seed=42):
    rnd = random.Random(seed)
    data = []

    for _, row in df_in.iterrows():
        drug = str(row["Drug_Name"]).strip() if not pd.isna(row["Drug_Name"]) else ""
        cancers = split_semicolon(row["Common_Cancers"])
        sidefx  = split_semicolon(row["Major_Side_Effects"])

        if not drug:
            continue

        cancers_text = ", ".join(cancers) if cancers else "cancer"
        sidefx_text  = "; ".join(sidefx) if sidefx else "side effects"

        template = rnd.choice(templates)
        text = template.replace("{DRUG}", drug)\
                       .replace("{CANCERS}", cancers_text)\
                       .replace("{SIDEFX}", sidefx_text)

        entities = []

        # DRUG
        for m in re.finditer(re.escape(drug), text):
            entities.append((m.start(), m.end(), "DRUG"))

        # CANCER
        for c in cancers:
            for m in re.finditer(re.escape(c), text):
                entities.append((m.start(), m.end(), "CANCER"))

        # SIDE_EFFECT
        for s in sidefx:
            for m in re.finditer(re.escape(s), text):
                entities.append((m.start(), m.end(), "SIDE_EFFECT"))

        entities = sorted(set(entities), key=lambda x: x[0])
        if entities:
            data.append((text, {"entities": entities}))

    return data

train_data = build_training_data(df_train, TEMPLATES, seed=42)
test_data  = build_training_data(df_test,  TEMPLATES, seed=99)

print("✅ Train samples:", len(train_data))
print("✅ Test samples :", len(test_data))
print("\n✅ Example train sample:\n", train_data[0])
```

Train spaCy NER Model:

```python
nlp = spacy.load("en_core_web_sm")
ner = nlp.get_pipe("ner")

for label in ["DRUG", "CANCER", "SIDE_EFFECT"]:
    ner.add_label(label)

optimizer = nlp.resume_training()

EPOCHS = 20
for epoch in range(EPOCHS):
    random.shuffle(train_data)
    losses = {}
    for text, annotations in train_data:
        example = Example.from_dict(nlp.make_doc(text), annotations)
        nlp.update([example], drop=0.3, sgd=optimizer, losses=losses)
    print(f"Epoch {epoch+1}/{EPOCHS} | NER Loss: {losses.get('ner')}")
```

Synthetic Test Evaluation (Sanity Check):

```python
def evaluate_ner(nlp, data):
    scorer = Scorer()
    examples = []
    for text, annotations in data:
        doc = nlp(text)
        example = Example.from_dict(doc, annotations)
        examples.append(example)
    return scorer.score(examples)

scores = evaluate_ner(nlp, test_data)

print("✅ SYNTHETIC METRICS (TEST SET)")
print("Precision:", round(scores["ents_p"], 3))
print("Recall   :", round(scores["ents_r"], 3))
print("F1 Score :", round(scores["ents_f"], 3))

print("\n✅ PER LABEL METRICS")
for label, s in scores["ents_per_type"].items():
    print(f"{label:12}  P={s['p']:.3f}  R={s['r']:.3f}  F1={s['f']:.3f}")
```

Extract Text from PDF:

```python
def extract_text_from_pdf(pdf_path):
    doc = fitz.open(pdf_path)
    pages = []
    for i in range(len(doc)):
        t = doc[i].get_text("text")
        if t:
            pages.append(t)
    doc.close()
    return "\n".join(pages)

pdf_text = extract_text_from_pdf(pdf_path)

print("✅ Extracted characters:", len(pdf_text))
print("\n--- SAMPLE (first 1000 chars) ---\n")
print(pdf_text[:1000])
```

Clean Extracted PDF Text:

```python
def clean_pdf_text(text):
    text = text.replace("\u00ad", "")      # soft hyphen
    text = re.sub(r"-\n", "", text)        # join hyphenated line breaks
    text = re.sub(r"\n+", "\n", text)      # collapse multiple newlines
    text = re.sub(r"[ \t]+", " ", text)
    return text

pdf_text_clean = clean_pdf_text(pdf_text)
print("✅ Cleaned characters:", len(pdf_text_clean))
```

 Build Knowledge Graph from PDF Entities:

```python
LABELS = {"DRUG", "CANCER", "SIDE_EFFECT"}

def build_kg_from_doc(doc):
    G = nx.Graph()

    for sent in doc.sents:
        ents = [(e.text.strip(), e.label_) for e in sent.ents if e.label_ in LABELS]

        drugs_s  = [t for t,l in ents if l == "DRUG"]
        cancers_s = [t for t,l in ents if l == "CANCER"]
        sidefx_s  = [t for t,l in ents if l == "SIDE_EFFECT"]

        for d in drugs_s:   G.add_node(d, node_type="DRUG")
        for c in cancers_s: G.add_node(c, node_type="CANCER")
        for s in sidefx_s:  G.add_node(s, node_type="SIDE_EFFECT")

        for d in drugs_s:
            for c in cancers_s:
                if G.has_edge(d, c):
                    G[d][c]["weight"] += 1
                else:
                    G.add_edge(d, c, relation="TREATS", weight=1)

        for d in drugs_s:
            for s in sidefx_s:
                if G.has_edge(d, s):
                    G[d][s]["weight"] += 1
                else:
                    G.add_edge(d, s, relation="CAUSES", weight=1)

    return G

G = build_kg_from_doc(doc_pdf)
print("✅ Knowledge Graph Created")
print("Nodes:", G.number_of_nodes())
print("Edges:", G.number_of_edges())
```

Visualize Knowledge graph  code

```python
# =========================
# OPTION 3 : TOP RELATIONS ONLY (AUTO DRAW)
# =========================

import matplotlib.pyplot as plt
import networkx as nx

if G.number_of_edges() == 0:
    print("❌ No edges found in graph.")
else:
    top_edges = 25

    edges_sorted = sorted(
        G.edges(data=True),
        key=lambda x: x[2].get("weight", 1),
        reverse=True
    )

    keep_edges = edges_sorted[:top_edges]

    H = nx.Graph()
    for u, v, data in keep_edges:
        H.add_node(u, **G.nodes[u])
        H.add_node(v, **G.nodes[v])
        H.add_edge(u, v, **data)

    pos = nx.spring_layout(H, seed=42, k=1.0)

    color_map = []
    for n in H.nodes():
        t = H.nodes[n].get("node_type")
        if t == "DRUG":
            color_map.append("#4e79a7")
        elif t == "CANCER":
            color_map.append("#e15759")
        elif t == "SIDE_EFFECT":
            color_map.append("#f28e2b")
        else:
            color_map.append("gray")

    widths = [1 + 0.6*H[u][v].get("weight", 1) for u,v in H.edges()]

    plt.figure(figsize=(14, 9))
    nx.draw_networkx_nodes(H, pos, node_color=color_map, node_size=850, alpha=0.9)
    nx.draw_networkx_edges(H, pos, width=widths, alpha=0.35)
    nx.draw_networkx_labels(H, pos, font_size=8)

    edge_labels = {(u, v): H[u][v].get("relation", "") for u, v in H.edges()}
    nx.draw_networkx_edge_labels(H, pos, edge_labels=edge_labels, font_size=7)

    plt.title("Top Drug–Cancer–Side Effect Relations", fontsize=14)
    plt.axis("off")
    plt.tight_layout()
    plt.show()

    print("✅ Top‑relation KG drawn | Nodes:", H.number_of_nodes(), "Edges:", H.number_of_edges())
```
